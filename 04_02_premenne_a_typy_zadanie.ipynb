# Premenné a dátové typy v Pythone

V tomto notebooku budeme objavovať, ako Python pracuje s rôznymi typmi dát a aké sú pravidlá pre pomenovanie premenných.

## Základná práca s premennými

Vyskúšajte si nasledujúce príkazy a pozorujte výstupy:

````python
# Vytvoríme premennú a
a = 7
print(a)
````

````python
# Skopírujeme hodnotu z a do b
b = a
print(b)
````

## Zistenie typu premennej

Python má funkciu `type()`, ktorá nám povie, aký typ dát premenná obsahuje.

````python
# Aký typ má premenná a?
type(a)
````

````python
# Aký typ má premenná b?
type(b)
````

### Rôzne dátové typy

Vyskúšajte si funkciu `type()` na rôznych hodnotách:

````python
# Celé číslo (integer)
type(58)
````

````python
# Desatinné číslo (float)
type(99.9)
````

````python
# Text - reťazec (string)
type('abc')
````

## Úvodzovky pri reťazcoch

V Pythone môžeme pre text použiť jednoduché `'text'` alebo dvojité `"text"` úvodzovky. Je to jedno!

````python
# Obe varianty sú správne
pozdrav1 = 'Ahoj'
pozdrav2 = "Ahoj"

print(pozdrav1)
print(pozdrav2)
print(type(pozdrav1))
print(type(pozdrav2))
````

## Pravidlá pre pomenovanie premenných

Nie všetky názvy premenných sú v Pythone povolené. Poďme si vyskúšať, ktoré názvy fungujú a ktoré nie.

### 🔒 Rezervované slová (keywords)

Python má **rezervované slová**, ktoré majú špeciálny význam v jazyku. **Nemožno ich použiť** ako názvy premenných!

Zoznam rezervovaných slov:

```
False      class      finally    is         return
None       continue   for        lambda     try
True       def        from       nonlocal   while
and        del        global     not        with
as         elif       if         or         yield
assert     else       import     pass
break      except     in         raise
```

Vyskúšajte si, čo sa stane, keď sa pokúsite použiť rezervované slovo ako názov premennej:

````python
# Pokus o použitie rezervovaného slova 'for'
for = 10
````

````python
# Pokus o použitie rezervovaného slova 'if'
if = 5
````

````python
# Pokus o použitie rezervovaného slova 'class'
class = "Python"
````

**Chybová hláška:** Python vám povie `SyntaxError: invalid syntax`, pretože tieto slová majú v Pythone špeciálny význam a nemožno ich použiť ako názvy premenných.

### ❌ Neplatné názvy premenných

Nasledujúce názvy **NEFUNGUJÚ** - vyskúšajte ich a pozrite sa na chybovú hlášku:

````python
# Názov začínajúci číslicou
1 = 5
````

````python
# Názov začínajúci číslicou
1a = 10
````

````python
# Názov začínajúci číslicou a podčiarkovníkom
1_ = 20
````

**Pravidlo:** Názov premennej **nesmie začínať číslicou**!

### ✅ Platné názvy premenných

Tieto názvy **FUNGUJÚ** správne:

````python
# Jednoduchý názov
a = 100
print(a)
````

````python
# Písmeno s číslicou (číslica nie je na začiatku)
a1 = 200
print(a1)
````

````python
# Podčiarkovníky sú povolené
a_b_c___95 = 300
print(a_b_c___95)
````

````python
# Názov môže začínať podčiarkovníkom
_abc = 400
print(_abc)
````

````python
# Podčiarkovník, potom číslica
_1a = 500
print(_1a)
````

### 🚫 Zlé názvy premenných (aj keď sú technicky povolené)

Niektoré názvy sú síce platné, ale **nie sú vhodné** na písanie čistého kódu:

````python
# ❌ Medzery nie sú povolené!
# number Of Users = 10  # Toto by spôsobilo chybu
````

````python
# ❌ Špeciálne znaky ako %% nie sú povolené!
# tep%%por = 5  # Toto by spôsobilo chybu
````

````python
# ⚠️ Tento názov je platný, ale nedáva zmysel
asdf = 42
print(asdf)
````

````python
# ⚠️ Diakritika nie je dobrý nápad (aj keď funguje)
používateľ = "Ján"
print(používateľ)
````

**Odporúčania pre dobré názvy premenných:**
- Používajte **zmysluplné názvy** (napr. `vek`, `meno`, `pocet_studentov`)
- Píšte názvy **po anglicky** (v praxi sa to často vyžaduje)
- **Bez diakritiky**
- Používajte **malé písmená** a podčiarkovníky (napr. `user_count`)
- Vyhnite sa skratkám, ktoré nedávajú zmysel

## Konštanty

Konštanty sú hodnoty, ktoré by sa **nemali meniť**. V Pythone je konvencia pomenovať konštanty **VEĽKÝMI PÍSMENAMI**:

````python
# Konštanta - mala by zostať nezmenená počas celého programu
FINAL_ANSWER = 42
print(FINAL_ANSWER)
````

````python
# Ďalšie príklady konštánt
PI = 3.14159
MAX_STUDENTS = 30
SCHOOL_NAME = "Naša škola"

print(PI)
print(MAX_STUDENTS)
print(SCHOOL_NAME)
````

**Poznámka:** Python technicky **nedokáže zabrániť** zmene konštanty, ale konvencia VEĽKÝCH PÍSMEN hovorí programátorom: "Túto hodnotu nemeň!"

````python
# Python vám dovolí zmeniť konštantu, ale NEMALI by ste to robiť!
FINAL_ANSWER = 100  # ⚠️ Technicky možné, ale ZLÝ nápad
print(FINAL_ANSWER)
````
