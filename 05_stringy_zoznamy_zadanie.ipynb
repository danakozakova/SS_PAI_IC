# Textové reťazce a zoznamy

V tomto notebooku sa naučíte pracovať s textom (strings) a zoznamami (lists) - dvoma veľmi dôležitými dátovými typmi v Pythone.

## Textové reťazce (Strings)

String je dátový typ na **ukladanie textu**. Už viete, že môžeme použiť jednoduché `'text'` alebo dvojité `"text"` úvodzovky.

### Vytváranie stringov

```python
# Jednoduché úvodzovky
meno = 'Anna'
print(meno)
```

```python
# Dvojité úvodzovky
mesto = "Bratislava"
print(mesto)
```

```python
# Oboje je string
print(type(meno))
print(type(mesto))
```

### Kedy použiť jednoduché a kedy dvojité úvodzovky?

Hodí sa to, keď chceme použiť úvodzovky **vnútri** textu:

```python
# Text obsahuje jednoduché úvodzovky (apostrof)
veta1 = "Toto je Peter's kniha"
print(veta1)
```

```python
# Text obsahuje dvojité úvodzovky
veta2 = 'Povedal: "Dobrý deň!"'
print(veta2)
```

### Viacriadkové stringy

Na dlhší text cez viac riadkov použijeme **trojité úvodzovky** `"""text"""`

```python
# Viacriadkový text
basen = """Rozkvitli
na záhradke
červené ružičky."""

print(basen)
```

### Špeciálne znaky - escape sekvencie

Niektoré znaky sa zapisujú pomocou **spätnej lomky** `\`

```python
# \n - nový riadok
text = "Prvý riadok\nDruhý riadok"
print(text)
```

```python
# \t - tabulátor (odsadenie)
text = "Meno:\tAnna\nVek:\t25"
print(text)
```

```python
text
```

```python
# \" - úvodzovky vnútri textu
text = "Povedal: \"Ahoj!\""
print(text)
```

```python
# \" - úvodzovky vnútri textu
text = 'Povedal: "Ahoj!"'
print(text)
```

### Raw stringy (r"...")

Ak nechceme, aby Python interpretoval `\`, použijeme **raw string** s predponou `r`

```python
# Bežný string - \n sa zmení na nový riadok
cesta1 = "C:\nova\slozka"
print(cesta1)
```

```python
# Raw string - \n zostane ako text
cesta2 = r"C:\nova\slozka"
print(cesta2)
```

## Operácie so stringami

### Spájanie stringov (konkatenácia)

```python
# Operátor + spája stringy
meno = "Anna"
priezvisko = "Nováková"
cele_meno = meno + " " + priezvisko
print(cele_meno)
```

```python
# Pozor: nemožno sčítať string a číslo!
vek = 25
text = "Vek: " + vek  # Toto spôsobí chybu!
```

```python
# Riešenie 1: previesť číslo na string pomocou str()
vek = 25
text = "Vek: " + str(vek)
print(text)
```

```python
# Riešenie 2: použiť čiarku v print() - automaticky pridá medzeru
vek = 25
print("Vek:", vek)
```

### Opakovanie stringov

```python
# Operátor * opakuje string
print("Ha" * 5)
print("=" * 20)
```

```python
# Praktické použitie - oddeľovacia čiara
print("Nadpis")
print("-" * 20)
print("Obsah")
```

### Dĺžka stringu - funkcia len()

```python
# Koľko znakov má text?
text = "Python"
dlzka = len(text)
print(dlzka)
```

```python
# Medzery sa tiež počítajú
veta = "Ahoj svet"
print(len(veta))
```

### Indexovanie - prístup k jednotlivým znakom

Každý znak v stringu má svoj **index** (pozíciu). Indexovanie začína od **0**!

```python
# P y t h o n
# 0 1 2 3 4 5

slovo = "Python"
print(slovo[0])  # Prvý znak
print(slovo[1])  # Druhý znak
print(slovo[5])  # Šiesty znak
```

```python
# Čo sa stane, keď skúsime index mimo rozsahu?
slovo = "Python"
print(slovo[10])  # Chyba!
```

### Negatívne indexy

Python umožňuje indexovanie **odzadu** pomocou záporných čísel:

```python
# P  y  t  h  o  n
# 0  1  2  3  4  5
#-6 -5 -4 -3 -2 -1

slovo = "Python"
print(slovo[-1])  # Posledný znak
print(slovo[-2])  # Predposledný znak
print(slovo[-6])  # Prvý znak (rovnaké ako [0])
```

### F-stringy (formátované stringy)

**F-stringy** sú moderný spôsob, ako vkladať premenné do textu. Pred úvodzovky dáme `f`:

```python
# Klasický spôsob (zložitejší)
meno = "Peter"
vek = 20
print("Volám sa " + meno + " a mám " + str(vek) + " rokov.")
```

```python
# F-string (jednoduchší a čitateľnejší)
meno = "Peter"
vek = 20
print(f"Volám sa {meno} a mám {vek} rokov.")
```

```python
# Môžeme vkladať aj výrazy
cena = 100
mnozstvo = 3
print(f"Celková cena: {cena * mnozstvo} €")
```

### Základné metódy stringov

```python
# upper() - prevedie na veľké písmená
text = "python"
print(text.upper())
```

```python
# lower() - prevedie na malé písmená
text = "PYTHON"
print(text.lower())
```

```python
# capitalize() - prvé písmeno veľké
text = "python je super"
print(text.capitalize())
```

```python
# strip() - odstráni medzery na začiatku a na konci
text = "   Python   "
print(f"'{text}'")
print(f"'{text.strip()}'")
```

**Dôležité:** Stringy sú v Pythone **immutable** (nemenné). Metódy nevracajú zmenený pôvodný string, ale vytvoria nový!

```python
# Ukážka immutability
text = "python"
text.upper()  # Toto nezmení pôvodný text!
print(text)   # Stále malé písmená
```

```python
# Správne: uložiť výsledok do premennej
text = "python"
text = text.upper()  # Priradíme novú hodnotu
print(text)
```

## Zoznamy (Lists)

**Zoznam (list)** je dátový typ, ktorý umožňuje **uložiť viac hodnôt** do jednej premennej.

### Vytváranie zoznamov

```python
# Zoznam čísel
cisla = [1, 2, 3, 4, 5]
print(cisla)
```

```python
# Zoznam textov
mena = ["Anna", "Peter", "Jana"]
print(mena)
```

```python
# Zoznam môže obsahovať rôzne typy dát
zmiesany = [1, "text", 3.14, True, None]
print(zmiesany)
```

```python
# Prázdny zoznam
prazdny = []
print(prazdny)
```

```python
# Typ?
print(type(cisla))
```

### Indexovanie zoznamov

Zoznamy fungujú podobne ako stringy - indexujeme od **0**:

```python
# ["Anna", "Peter", "Jana"]
#    0        1        2

mena = ["Anna", "Peter", "Jana"]
print(mena[0])  # Prvý prvok
print(mena[1])  # Druhý prvok
print(mena[2])  # Tretí prvok
```

```python
# Čo sa stane pri indexe mimo rozsahu?
mena = ["Anna", "Peter", "Jana"]
print(mena[5])  # Chyba IndexError!
```

### Negatívne indexy pri zoznamoch

```python
# ["Anna", "Peter", "Jana"]
#    0        1        2
#   -3       -2       -1

mena = ["Anna", "Peter", "Jana"]
print(mena[-1])  # Posledný prvok
print(mena[-2])  # Predposledný prvok
print(mena[-3])  # Prvý prvok
```

### Zmena prvkov zoznamu

Na rozdiel od stringov sú zoznamy **mutable** (meniteľné) - môžeme meniť ich prvky!

```python
# Zmena prvku na konkrétnom indexe
mena = ["Anna", "Peter", "Jana"]
print("Pred zmenou:", mena)

mena[1] = "Pavol"  # Zmeníme druhý prvok
print("Po zmene:", mena)
```

### Dĺžka zoznamu - funkcia len()

```python
# Koľko prvkov má zoznam?
cisla = [10, 20, 30, 40, 50]
print(len(cisla))
```

```python
# Dĺžka prázdneho zoznamu
prazdny = []
print(len(prazdny))
```

## Operácie so zoznamami

### Spájanie zoznamov

```python
# Operátor + spája zoznamy
zoznam1 = [1, 2, 3, 4]
zoznam2 = [4, 5, 6]
vysledok = zoznam1 + zoznam2
print(vysledok)
```

### Opakovanie zoznamov

```python
# Operátor * opakuje zoznam
zoznam = [1, 2, 3]
vysledok = zoznam * 3
print(vysledok)
```

```python
# Praktické použitie - vytvorenie zoznamu so samými nulami
nuly = [0] * 10
print(nuly)
```

### Pridávanie prvkov do zoznamu

#### Metóda append() - pridá prvok na koniec

```python
# append() pridá JEDEN prvok na koniec zoznamu
cisla = [1, 2, 3]
print("Pred:", cisla)

cisla.append(4)
print("Po append(4):", cisla)

cisla.append(5)
print("Po append(5):", cisla)
```

```python
# Čo sa stane, keď pridáme zoznam?
cisla = [1, 2, 3]
cisla.append([4, 5])
print(cisla)  # Pridá sa celý zoznam ako jeden prvok!
```

#### Metóda extend() - pridá viac prvkov

```python
# extend() pridá všetky prvky z iného zoznamu
cisla = [1, 2, 3]
print("Pred:", cisla)

cisla.extend([4, 5, 6])
print("Po extend([4, 5, 6]):", cisla)
```

```python
# Rozdiel medzi append() a extend()
zoznam1 = [1, 2, 3]
zoznam1.append([4, 5])
print("S append():", zoznam1)

zoznam2 = [1, 2, 3]
zoznam2.extend([4, 5])
print("S extend():", zoznam2)
```

#### Metóda insert() - vloží prvok na konkrétnu pozíciu

```python
# insert(index, hodnota) vloží hodnotu na daný index
cisla = [1, 2, 3, 4]
print("Pred:", cisla)

cisla.insert(2, 99)  # Vloží 99 na index 2
print("Po insert(2, 99):", cisla)
```

### Odoberanie prvkov zo zoznamu

#### Metóda remove() - odoberie prvý výskyt hodnoty

```python
# remove(hodnota) odoberie PRVÝ výskyt danej hodnoty
ovocie = ["jablko", "hruška", "banán", "jablko"]
print("Pred:", ovocie)
```

```python
ovocie.remove("jablko")
print("Po remove('jablko'):", ovocie)
```

#### Metóda pop() - odoberie prvok na danom indexe

```python
# pop(index) odoberie prvok na indexe a vráti ho
cisla = [10, 20, 30, 40, 50]
print("Pred:", cisla)
```

```python
odobrany = cisla.pop(0)  # Odoberie prvok na indexe 2
print("Odobraný prvok:", odobrany)
print("Po pop(0):", cisla)
```

```python
odobrany = cisla.pop(0)  # Odoberie prvok na indexe 2
print("Odobraný prvok:", odobrany)
print("Po pop(0):", cisla)
```

```python
odobrany = cisla.pop(0)  # Odoberie prvok na indexe 2
print("Odobraný prvok:", odobrany)
print("Po pop(0):", cisla)
```

```python
# pop() bez argumentu odoberie posledný prvok
cisla = [10, 20, 30, 40, 50]
posledny = cisla.pop()
print("Posledný prvok:", posledny)
print("Zvyšok:", cisla)
```

### Kontrola, či zoznam obsahuje prvok

```python
# Operátor 'in' kontroluje, či prvok je v zozname
ovocie = ["jablko", "hruška", "banán"]

print("jablko" in ovocie)
print("pomaranč" in ovocie)
```

```python
# 'not in' kontroluje, či prvok NIE JE v zozname
print("melón" not in ovocie)
```

## Zhrnutie

V tomto notebooku ste sa naučili:

### Stringy:
- Vytváranie stringov s `'...'` alebo `"..."`
- Viacriadkové stringy s `"""..."""`
- Escape sekvencie: `\n`, `\t`, `\"`
- Raw stringy: `r"..."`
- Operácie: `+` (spájanie), `*` (opakovanie)
- Funkcia `len()` pre dĺžku
- Indexovanie (od 0, negatívne indexy)
- F-stringy: `f"text {premenna}"`
- Metódy: `.upper()`, `.lower()`, `.capitalize()`, `.strip()`
- Stringy sú **immutable** (nemenné)

### Zoznamy:
- Vytváranie: `[prvok1, prvok2, ...]`
- Indexovanie (od 0, negatívne indexy)
- Zoznamy sú **mutable** (meniteľné)
- Operácie: `+` (spájanie), `*` (opakovanie)
- Funkcia `len()` pre dĺžku
- Metódy pridávania: `.append()`, `.extend()`, `.insert()`
- Metódy odoberania: `.remove()`, `.pop()`
- Operátor `in` na kontrolu prítomnosti prvku
