## Cvičenie 1: Prihlasovacie meno

Napíš funkciu `login(meno, priezvisko, rok)`, ktorá vytvorí prihlasovacie meno zložené z:
- prvých troch písmen mena, malými písmenami
- celého priezviska, malými písmenami
- posledných dvoch číslic roku

Príklad:
```python
login('Jana', 'Novakova', 2007)
```

Výsledok:
```
jannovakova07
```

Pomôcka: číslo preveď na text cez `str(...)`, časť reťazca získaš rezom (napr. `text[:3]`, `text[-2:]`), malé písmená cez `.lower()`.

````python
# Sem napíšte funkciu "login"
````

### Test cvičenia 1:

````python
if login('Jana', 'Novakova', 2007) == 'jannovakova07':
    print('Passed!')
else:
    print('Pre argumenty "Jana", "Novakova", 2007 sa očakáva jannovakova07')
    print('Vaša funkcia vrátila:', login('Jana', 'Novakova', 2007))

if login('Adam', 'Biely', 2010) == 'adabiely10':
    print('Passed!')
else:
    print('Pre argumenty "Adam", "Biely", 2010 sa očakáva adabiely10')
    print('Vaša funkcia vrátila:', login('Adam', 'Biely', 2010))
````

## Cvičenie 2: Počet rôznych hodnôt

Napíš funkciu `pocet_roznych(polozky)`, ktorá dostane zoznam a vráti, koľko **rôznych (unikátnych)** hodnôt obsahuje.

Príklad:
```python
pocet_roznych(['a', 'b', 'a', 'c', 'b'])
```

Výsledok:
```
3
```

Pomôcka: spomeň si, čím sa líši **množina** od zoznamu a akou funkciou zistíš počet prvkov.

````python
# Sem napíšte funkciu "pocet_roznych"
````

### Test cvičenia 2:

````python
if pocet_roznych(['a', 'b', 'a', 'c', 'b']) == 3:
    print('Passed!')
else:
    print('Pre zoznam ["a","b","a","c","b"] sa očakáva 3')
    print('Vaša funkcia vrátila:', pocet_roznych(['a', 'b', 'a', 'c', 'b']))

if pocet_roznych([]) == 0:
    print('Passed!')
else:
    print('Pre prázdny zoznam sa očakáva 0')
    print('Vaša funkcia vrátila:', pocet_roznych([]))
````

## Cvičenie 3: Úprava mena

Napíš funkciu `uprav_meno(meno)`, ktorá odstráni medzery na okrajoch a vráti meno s **veľkým prvým písmenom** a ostatnými malými.

Príklad:
```python
uprav_meno('  jANA  ')
```

Výsledok:
```
Jana
```

Pomôcka: metódy sa dajú reťaziť – každá pracuje s výsledkom predošlej (napr. `text.strip().capitalize()`).

````python
# Sem napíšte funkciu "uprav_meno"
````

### Test cvičenia 3:

````python
if uprav_meno('  jANA  ') == 'Jana':
    print('Passed!')
else:
    print('Pre "  jANA  " sa očakáva Jana')
    print('Vaša funkcia vrátila:', uprav_meno('  jANA  '))

if uprav_meno('peter') == 'Peter':
    print('Passed!')
else:
    print('Pre "peter" sa očakáva Peter')
    print('Vaša funkcia vrátila:', uprav_meno('peter'))
````

## Cvičenie 4 (BONUS): Spoločné hodnoty

Napíš funkciu `spolocne(zoznam1, zoznam2)`, ktorá vráti, koľko hodnôt sa nachádza **v oboch** zoznamoch naraz.

Príklad:
```python
spolocne(['a', 'b', 'c'], ['b', 'c', 'd'])
```

Výsledok:
```
2
```

Pomôcka: prevedením zoznamu na **množinu** a použitím **prieniku** (`&`) dostaneš spoločné hodnoty.

````python
# Sem napíšte funkciu "spolocne"
````

### Test cvičenia 4:

````python
if spolocne(['a', 'b', 'c'], ['b', 'c', 'd']) == 2:
    print('Passed!')
else:
    print('Pre ["a","b","c"] a ["b","c","d"] sa očakáva 2')
    print('Vaša funkcia vrátila:', spolocne(['a', 'b', 'c'], ['b', 'c', 'd']))

if spolocne([1, 2, 3], [4, 5]) == 0:
    print('Passed!')
else:
    print('Pre [1,2,3] a [4,5] sa očakáva 0')
    print('Vaša funkcia vrátila:', spolocne([1, 2, 3], [4, 5]))
````
