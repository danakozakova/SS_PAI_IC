## Cvičenie 1: Prihlasovacie meno

Napíš funkciu `login(meno, priezvisko, rok)`, ktorá vytvorí prihlasovacie meno zložené z:
- prvých dvoch písmen mena, malými písmenami
- prvých štyroch písmen priezviska, malými písmenami
- posledných dvoch číslic roku

Príklad:
```python
login('Jana', 'Novakova', 2007)
```

Výsledok:
```
janova07
```

Pomôcka: číslo preveď na text cez `str(...)`, časť reťazca získaš rezom (napr. `text[:2]`, `text[-2:]`), malé písmená cez `.lower()`.

````python
# Sem napíšte funkciu "login"
````

### Test cvičenia 1:

````python
if login('Jana', 'Novakova', 2007) == 'janova07':
    print('Passed!')
else:
    print('Pre argumenty "Jana", "Novakova", 2007 sa očakáva janova07')
    print('Vaša funkcia vrátila:', login('Jana', 'Novakova', 2007))

if login('Adam', 'Biely', 2010) == 'adbiel10':
    print('Passed!')
else:
    print('Pre argumenty "Adam", "Biely", 2010 sa očakáva adbiel10')
    print('Vaša funkcia vrátila:', login('Adam', 'Biely', 2010))
````

## Cvičenie 2: Počet rôznych hodnôt

Napíš funkciu `pocet_roznych(polozky)`, ktorá dostane zoznam a vráti, koľko **rôznych (unikátnych)** hodnôt obsahuje.

Príklad:
```python
pocet_roznych([1, 2, 2, 3, 3, 3])
```

Výsledok:
```
3
```

Pomôcka: spomeň si, čím sa líši **množina** od zoznamu a akou funkciou zistíš počet prvkov.

````python
# Sem napíšte funkciu "pocet_roznych"
````

### Test cvičenia 2:

````python
if pocet_roznych([1, 2, 2, 3, 3, 3]) == 3:
    print('Passed!')
else:
    print('Pre zoznam [1,2,2,3,3,3] sa očakáva 3')
    print('Vaša funkcia vrátila:', pocet_roznych([1, 2, 2, 3, 3, 3]))

if pocet_roznych([]) == 0:
    print('Passed!')
else:
    print('Pre prázdny zoznam sa očakáva 0')
    print('Vaša funkcia vrátila:', pocet_roznych([]))
````

## Cvičenie 3: Úprava mena

Napíš funkciu `uprav_meno(meno)`, ktorá odstráni medzery na okrajoch a vráti meno **veľkými písmenami**.

Príklad:
```python
uprav_meno('  jana  ')
```

Výsledok:
```
JANA
```

Pomôcka: metódy sa dajú reťaziť – každá pracuje s výsledkom predošlej (napr. `text.strip().upper()`).

````python
# Sem napíšte funkciu "uprav_meno"
````

### Test cvičenia 3:

````python
if uprav_meno('  jana  ') == 'JANA':
    print('Passed!')
else:
    print('Pre "  jana  " sa očakáva JANA')
    print('Vaša funkcia vrátila:', uprav_meno('  jana  '))

if uprav_meno('Peter') == 'PETER':
    print('Passed!')
else:
    print('Pre "Peter" sa očakáva PETER')
    print('Vaša funkcia vrátila:', uprav_meno('Peter'))
````

## Cvičenie 4 (BONUS): Hodnoty iba v prvom zozname

Napíš funkciu `iba_v_prvom(zoznam1, zoznam2)`, ktorá vráti, koľko hodnôt je **iba v prvom** zozname a v druhom sa nenachádzajú.

Príklad:
```python
iba_v_prvom(['a', 'b', 'c'], ['b', 'c', 'd'])
```

Výsledok:
```
1
```

Pomôcka: prevedením zoznamov na **množiny** a použitím **rozdielu** (`-`) dostaneš hodnoty iba z prvej množiny.

````python
# Sem napíšte funkciu "iba_v_prvom"
````

### Test cvičenia 4:

````python
if iba_v_prvom(['a', 'b', 'c'], ['b', 'c', 'd']) == 1:
    print('Passed!')
else:
    print('Pre ["a","b","c"] a ["b","c","d"] sa očakáva 1')
    print('Vaša funkcia vrátila:', iba_v_prvom(['a', 'b', 'c'], ['b', 'c', 'd']))

if iba_v_prvom([1, 2], [1, 2, 3]) == 0:
    print('Passed!')
else:
    print('Pre [1,2] a [1,2,3] sa očakáva 0')
    print('Vaša funkcia vrátila:', iba_v_prvom([1, 2], [1, 2, 3]))
````
